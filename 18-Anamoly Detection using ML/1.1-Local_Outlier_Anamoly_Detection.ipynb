```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.neighbors import LocalOutlierFactor
```

```python
X, y = make_circles(n_samples=500, factor=0.5, noise=0.05)
```

```python
# Add artificial outliers
rng = np.random.RandomState(42)
outliers = rng.uniform(low=-1.5, high=1.5, size=(30, 2))
X_all = np.vstack([X, outliers])
```

```python
# Plot
plt.scatter(X_all[:, 0], X_all[:, 1], s=15)
```

```python
# Apply LOF
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.06)
```

```python
y_pred = lof.fit_predict(X_all)
```

```python
plt.scatter(X_all[:, 0], X_all[:, 1], c=y_pred, s=15)
```
