```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(n_samples=500, centers=3, cluster_std=0.6, random_state=42)
```

```python
# Add artificial outliers
rng = np.random.RandomState(42)
outliers = rng.uniform(low=-10, high=10, size=(30, 2))
X_all = np.vstack([X, outliers])
```

```python
plt.scatter(X_all[:, 0], X_all[:, 1], s=15)
```

```python
from sklearn.ensemble import IsolationForest
```

```python
iso = IsolationForest(n_estimators=100, contamination=0.06, random_state=42)
```

```python
y_pred = iso.fit_predict(X_all)
```

```python
plt.scatter(X_all[:, 0], X_all[:, 1], c=y_pred, s=15)
```
